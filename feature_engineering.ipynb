```python
import pandas as pd
df = pd.read_csv("data/cleaned_products.csv")
```

```python
for i in df.product_title:
    print(i, "\n\n")
```

```python
df["aux_col"] = df.product_title.str.split(" ")
df["special_name"] = df.aux_col.str[0]
df = df.drop(["aux_col"], axis = 1)
```

```python
df["product_code"] = ((df.product_code.str.split("-")).str[0]).fillna("0")

df = df.rename(columns = {"product_code" : "product_family"})
```
