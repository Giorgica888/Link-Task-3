```python
import pandas as pd 
```

```python
url = "https://raw.githubusercontent.com/Giorgica888/Link-Task-3/refs/heads/main/data/products.csv"
df = pd.read_csv(url)
```

```python
df
```

```python
def clean_columns(df):
    df = df.copy()

    columns = list(df.columns)
    new_columns = []
    for i in columns:
        i = i.strip().strip("_")
        i = i.lower()
        i = i.replace(" ", "_")
        new_columns.append(i)

    df.columns = new_columns
    return df

def lower_the_title(df):
    df = df.copy()
    df.product_title = df.product_title.str.lower()
    return df

def delete_where_title_and_target_are_missing(df):
    df = df.copy()
    df = df.dropna(subset = ["product_title"])
    return df.dropna(subset = ["category_label"])

def delete_duplicates(df):
    df = df.copy()
    return df.drop_duplicates(["product_title", "category_label"])

def replace_some_names(df):
    df = df.copy()
    df.category_label = df.category_label.replace({
    "Mobile Phone" : "Mobile Phones",
    "CPU" : "CPUs",
    "fridge" : "Fridges",
    "Fridge Freezers" : "Fridges",
    })
    return df

def delete_useless_columns(df):
    df = df.copy()
    return df.drop(["listing_date", "merchant_rating", "product_id", "number_of_views", "listing_date", 'merchant_id', 'product_code'], axis = 1)

def lower_names(df):
    df = df.copy()
    df.category_label = df.category_label.str.lower()
    df.category_label = df.category_label.str.replace(" ", "_")
    return df


def categorise_columns(df):
    df = df.copy()
    df.category_label = df.category_label.astype("category")
    df.merchant_id = df.merchant_id.astype("category")
    return df

def reset_the_index(df):
    df = df.copy()
    return df.reset_index(drop = True)

def pipeline(df):
    return (
        df.pipe(clean_columns)
        .pipe(lower_the_title)
        .pipe(delete_where_title_and_target_are_missing)
        .pipe(delete_duplicates)
        .pipe(replace_some_names)
        .pipe(lower_names)
        .pipe(categorise_columns)
        .pipe(delete_useless_columns)

    )
```

```python
df = df.pipe(pipeline)
```
